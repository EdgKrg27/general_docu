# Acciones

```python
# collect()
rdd = sc.parallelize([1,2,3,4,5,6])
rdd.collect()
```

```python
# coutn()
rdd.count()
```

```python
# take()
rdd.take(2)
```

```python
# first()
rdd.first()
```

```python
# top(n , key=None)
rdd = sc.parallelize([10,4,25,7,2])
rdd.top(3)
```

```python
# reduce(func)
rdd.reduce(lambda x, y: x + y)
```

```python
# countByKey()
rdd = sc.parallelize([("Ventas", 1), ("IT", 1), ("Ventas", 1)])
rdd.countByKey()
```

# Ejercicios

```python
# Cree un RDD llamado importes a partir del archivo adjunto
importes = sc.textFile("num.txt")
```

```python
# ¿Cuántos registros tiene el RDD importes?
importes.count()
```

```python
# cuál es el valor mínimo y máximo del RDD importes
min_importes = importes.min()
max_importes = importes.max()
print(min_importes)
print(max_importes)
```

```python
# cree un RDD top15 que contenga los 15 mayores valores del RDD importes, tenga en cuenta que pueden repetirse los valroes.
top15 = importes.top(15)
```

```python
# Escribir top15 en archivo salida
salida = sc.parallelize(top15)
salida.saveAsTextFile("salida")
```

```python
# Cree una función llamada factorial que calcule el factorial de un número dado como parámetro
def factorial(n):
  if n == 0:
    return 1
  else:
    rdd = sc.parallelize(list(range(1, n+1)))
    return rdd.reduce(lambda x, y: x * y)
```

```python
factorial(0)
```

```python
factorial(1)
```

```python
factorial(2)
```
