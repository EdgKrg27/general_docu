# Variables

Las variables permiten asignar nombres coherentes a información para ser reutilizada con facilidad.  
La forma de dar valor a una variable se realiza por asignaciones y en Python puede ser simples o múltiples, todas las asignaciones en Python se realizan utilizando el carácter `=` y la dirección es hacia la izquierda, por tanto, se guarda en la parte izquiera lo que se declare en la parte derecha. Por ejemplo:

`identificador = [literal o expresión]`


## Asignaciones simples

Las asignaciones simples son cuando se asigna solamente un solo valor a una única variable

```python
str_color = 'Red'
n_cards = 10
pi = 3.1416
buy_list = ['Bread', 'Oil']
locations = {"madrid":(40.23,3.42), "dublin":(53.0, 6.15)}
```

## Asignaciones multiples

Las asignaciones múltiples es cuando se asigna varios valores a varias variables a la vez.

```python
str_red, str_blue = 'red', 'blue'
pi_number, g_force = [3.1416, 9.8]
ltr1, ltr2, ltr3 = 'red'
```

## Literales y expresiones

- **Expresiones**: son datos asignados tras la evaluación de una o varias sentencias. Suelen ser dinámicos y no se concoe el valor final analizando el código

## Nomenclatura

Segun la PEP8 los nombres de variables se debene escribir en formato `snake_case`, tomando en cuenta que también se deben cumplir las siguientes características:

- Tiene que empezar por una letra o barra baja
- El uso de keywords como nombres está prohibido
- Los nombres deben ser descriptivos
- Deben estar en minúsculas y separando palabras pr barras bajas `_`
- Las constantes se escriben en mayúsculas y SNAKE_CASE

## Tipos de datos

Definen que clase de información estamos manejando. Existen varios tipos de datos en python y en todos los lenguajes de programación, lo más utilizados son:

- Enteros (int): Números sin decimales, como 5 o -10
- Flotantes (float): Números con decimales como 3.14, 2.0
- Cadenas (str): Texto encerrado en comillas, como "Hola Mundo"
- Booleanos (bool): VAlores de lógica que solo pueden ser True o False

En python se conoce como tipo de dato dinámico, a diferencia de otros lenguajes de programación donde una variable nace con un tipo y muere con ese mismo tipo, en Python las variables son nombres que aputan a *"objetos"* de memoria.

```python
print(type(5))  # entero
print(type(256745674567.5234353463546))  # flotante
print(type("Hola Mundo"))  # string o cadena de texto
print(type(True))  # bool
```

Se pueden realizar operaciones con estos datos, pero también definen que tipo de operaciones son permitidas. Aunque es dinámico por los objetos, Python es fuertemente tipado, es decir, no se pude mezclar tipos de datos no compatibles

```python
# esta operación si se puede, a pesar que son dos tipos de datos diferentes, son compatibles
print(5 + 7.3)  # int + float
```

```python
# esta operación no se permite ya que los tipos de datos no son compatibles
print("7" + 3)  # str + int
```

ya que python se necesita ser explisito con las operaciones, para solucionar el problema de querer sumar un string con un número cual sea, se necesita hacer la converisón de tipo de dato del string, cabe mencionar que solamente la conversión de tipo de dato funcionara si este es posible en caso contrario retornara un error de conversión.

Para transformar datos, usamos funciones que tienen el mismo nombre que el tipo de dato:

- `str()` convierte a texto
- `int()` convierte a entero
- `float()` convierte a flotante

```python
print(int("7") + 3)
```

## Operadores de asignación

Estos actuan directamente sobre la variable actual modificando su valor

```python
# Suma en asignación
a = 5
a += 5
print(a)
```

```python
# Multiplicación en asignación
a = 5
a *= 5
print(a)
```

```python
# Resta en asignación
a = 5
a -= 5
print(a)
```

```python
# División en asignación
a = 5
a /= 5
print(a)
```

```python
# Modulo en asignación
a = 5
a %= 2
print(a)
```

```python
# Potencia en asignacilón
a = 5
a **= 3
print(a)
```
